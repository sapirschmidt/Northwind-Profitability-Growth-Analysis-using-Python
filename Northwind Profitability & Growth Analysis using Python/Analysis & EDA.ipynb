## יצירת חישובי Profitabilty ומטריקות

```python
profitability = orders.merge(
    order_details,
    on="OrderID",
    how="inner"
)
profitability = profitability.merge(
    products,
    on="ProductID",
    how="inner",
    suffixes=("_order", "_product")
)
```

```python
profitability = profitability.merge(
    categories[["CategoryID", "CategoryName"]],
    on="CategoryID",
    how="left"
)
```

```python
profitability["BuyingCost"] = profitability["UnitPrice_product"]
profitability["SellingPrice"] = profitability["UnitPrice_order"]

profitability["Revenue"] = (
    profitability["Quantity"] *
    profitability["SellingPrice"] *
    (1 - profitability["Discount"])
)

profitability["COGS"] = (
    profitability["Quantity"] *
    profitability["BuyingCost"]
)

profitability["GrossProfit"] = (
    profitability["Revenue"] -
    profitability["COGS"]
)

profitability["GrossMargin"] = np.where(
    profitability["Revenue"] != 0,
    profitability["GrossProfit"] /
    profitability["Revenue"],
    np.nan
)
```

## Which products and categories generate the highest revenue and gross profit, and which products have weak profitability?

```python
product_profitability = (
    profitability
    .groupby(
        ["ProductID", "ProductName"],
        as_index=False
    )
    .agg(
        UnitsSold=("Quantity", "sum"),
        Revenue=("Revenue", "sum"),
        COGS=("COGS", "sum"),
        GrossProfit=("GrossProfit", "sum")
    )
)

product_profitability["GrossMargin"] = np.where(
    product_profitability["Revenue"] != 0,
    product_profitability["GrossProfit"] /
    product_profitability["Revenue"],
    np.nan
)

product_profitability["ProfitPerUnit"] = np.where(
    product_profitability["UnitsSold"] != 0,
    product_profitability["GrossProfit"] /
    product_profitability["UnitsSold"],
    np.nan
)
```

## Which high-volume products generate disproportionately low gross profit or gross margin?

```python
volume_margin = product_profitability[
    [
        "ProductID",
        "ProductName",
        "UnitsSold",
        "Revenue",
        "GrossProfit",
        "GrossMargin"
    ]
].copy()
```

```python
volume_margin["VolumeRank"] = (
    volume_margin["UnitsSold"]
    .rank(ascending=False)
)

volume_margin["MarginRank"] = (
    volume_margin["GrossMargin"]
    .rank(ascending=False)
)

volume_threshold = volume_margin["UnitsSold"].median()
margin_threshold = volume_margin["GrossMargin"].median()

volume_margin["Segment"] = np.select(
    [
        (
            (volume_margin["UnitsSold"] >= volume_threshold) &
            (volume_margin["GrossMargin"] >= margin_threshold)
        ),
        (
            (volume_margin["UnitsSold"] >= volume_threshold) &
            (volume_margin["GrossMargin"] < margin_threshold)
        ),
        (
            (volume_margin["UnitsSold"] < volume_threshold) &
            (volume_margin["GrossMargin"] >= margin_threshold)
        )
    ],
    [
        "High Volume / High Margin",
        "High Volume / Low Margin",
        "Low Volume / High Margin"
    ],
    default="Low Volume / Low Margin"
)

volume_margin[
    volume_margin["Segment"] ==
    "High Volume / Low Margin"
]
```

## What are the main drivers of changes in Northwind's gross profit over time?

## MoM Growth

```python
profitability["Month"] = (
    profitability["OrderDate"]
    .dt.to_period("M")
)

monthly_profitability = (
    profitability
    .groupby("Month", as_index=False)
    .agg(
        Revenue=("Revenue", "sum"),
        COGS=("COGS", "sum"),
        GrossProfit=("GrossProfit", "sum"),
        UnitsSold=("Quantity", "sum"),
        Orders=("OrderID", "nunique")
    )
)

monthly_profitability["GrossMargin"] = (
    monthly_profitability["GrossProfit"] /
    monthly_profitability["Revenue"]
)

monthly_profitability["RevenueGrowth"] = (
    monthly_profitability["Revenue"]
    .pct_change()
)

monthly_profitability["ProfitGrowth"] = (
    monthly_profitability["GrossProfit"]
    .pct_change()
)
```

## Discount Analysis

```python
profitability["DiscountBand"] = np.select(
    [
        profitability["Discount"] == 0,
        profitability["Discount"] <= 0.05,
        profitability["Discount"] <= 0.10,
        profitability["Discount"] <= 0.20
    ],
    [
        "0%",
        "0-5%",
        "5-10%",
        "10-20%"
    ],
    default="20%+"
)

discount_analysis = (
    profitability
    .groupby("DiscountBand", as_index=False)
    .agg(
        Orders=("OrderID", "nunique"),
        UnitsSold=("Quantity", "sum"),
        Revenue=("Revenue", "sum"),
        GrossProfit=("GrossProfit", "sum")
    )
)

discount_analysis["GrossMargin"] = (
    discount_analysis["GrossProfit"] /
    discount_analysis["Revenue"]
)
```

## RFM CUSTOMER ANALYSIS

```python
# ==========================================
# RFM CUSTOMER ANALYSIS
# ==========================================

# Reference date = one day after the last order in the dataset
reference_date = profitability["OrderDate"].max() + pd.Timedelta(days=1)

rfm = (
    profitability
    .groupby("CustomerID")
    .agg(
        LastPurchaseDate=("OrderDate", "max"),
        Frequency=("OrderID", "nunique"),
        Monetary=("Revenue", "sum"),
        GrossProfit=("GrossProfit", "sum")
    )
    .reset_index()
)

# Recency = days since last purchase
rfm["Recency"] = (
    reference_date - rfm["LastPurchaseDate"]
).dt.days

# Reorder columns
rfm = rfm[
    [
        "CustomerID",
        "LastPurchaseDate",
        "Recency",
        "Frequency",
        "Monetary",
        "GrossProfit"
    ]
]

print(rfm.head())

# ==========================================
# RFM SCORING
# ==========================================

rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5, 4, 3, 2, 1],
    duplicates="drop"
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(int).astype(str)
    + rfm["F_Score"].astype(int).astype(str)
    + rfm["M_Score"].astype(int).astype(str)
)

rfm["RFM_Total"] = (
    rfm["R_Score"].astype(int)
    + rfm["F_Score"].astype(int)
    + rfm["M_Score"].astype(int)
)

print(rfm.head())

# ==========================================
# RFM CUSTOMER SEGMENTS
# ==========================================

rfm["RFM_Segment"] = np.select(
    [
        rfm["RFM_Total"] >= 13,
        rfm["RFM_Total"] >= 10,
        rfm["RFM_Total"] >= 7,
        rfm["RFM_Total"] >= 4
    ],
    [
        "Champions",
        "Loyal Customers",
        "Potential Loyalists",
        "At Risk"
    ],
    default="Lost / Low Value"
)

print(
    rfm[
        [
            "CustomerID",
            "Recency",
            "Frequency",
            "Monetary",
            "GrossProfit",
            "RFM_Score",
            "RFM_Total",
            "RFM_Segment"
        ]
    ].head(20)
)

rfm_summary = (
    rfm
    .groupby("RFM_Segment", as_index=False)
    .agg(
        Customers=("CustomerID", "nunique"),
        Revenue=("Monetary", "sum"),
        GrossProfit=("GrossProfit", "sum"),
        AvgRecency=("Recency", "mean"),
        AvgFrequency=("Frequency", "mean"),
        AvgMonetary=("Monetary", "mean")
    )
)

rfm_summary["GrossMargin"] = np.where(
    rfm_summary["Revenue"] != 0,
    rfm_summary["GrossProfit"] / rfm_summary["Revenue"],
    np.nan
)

print(rfm_summary)
```
